## Task- 2 **Speech to Text (Whisper)**

```python
import whisper
import pandas as pd
import nltk
from nltk.tokenize import sent_tokenize

# Download sentence tokenizer model (only first time)
nltk.download('punkt')
```

```python
# Load the medium Whisper model
model = whisper.load_model("medium")  # medium whisper model

# Path to your audio file
audio_file = r"D:\Year_2ADS_AI\DSAI_2_BLOCK-A\fae2-nlpr-group-group-11-1\speech_to_text_whisper\Is This Really The Best Pizza In Denver  Full Episode S7 E2  Kitchen Nightmares  Gordon Ramsay.mp3"

# Transcribe the audio
result = model.transcribe(audio_file, language="en")  # change language if needed

# Extract the text
text = result['text']

# Optional: clean text
text_clean = text.encode('utf-8', errors='ignore').decode('utf-8')

# Split text into sentences
sentences = sent_tokenize(text_clean)


# Print first 5 sentences to check
print("# Sample sentences (first 5):")
print("[")
for s in sentences[:5]:
    print(f"  '{s}',")
print("]")
```

```python
# Print first 15 sentences to check
print("# Sample sentences (first 15):")
print("[")
for s in sentences[:15]:
    print(f"  '{s}',")
print("]")
```

### IV. Saving Sentence to csv (Medium Whisper model)

```python
# Save sentences to CSV
df = pd.DataFrame(sentences, columns=['Sentence'])
df.to_csv("transcribed_data_whisper.csv", index=False)
```
